```python
import torch
import torch.nn as nn

def make_ffn(C):
  return nn.Sequential(nn.Linear(C , 4 * 8), nnReLU() , nn.Linear(4*C, C))
class EncoderBlock(nn.Module):
  def __init__(self, C , H):
    super().__init__()
    self.attn = nn.MultheadAttention(C , H , batch_first= True)
    self.ffn = make_ffn(C)
    self.ln1 = nn.LayerNorm(C)
    self.ln2 = nn.LayerNorm(C)

  def forward(self, x ):
    out, _ = self.attn(x , x ,x )
    x = self.ln1(x + out)
    x = self.ln2(x + self.ffn(x))
    return x
```
