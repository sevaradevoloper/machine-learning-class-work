```python
import pandas as pd
import numpy as np

# 1. Dataset yaratamiz
data = {
    'maydon': [50, 60, np.nan, 80, 50, 500],
    'narx': [400, 500, 550, 700, 400, 9500]
}
df = pd.DataFrame(data)
df
```

```python

# 2. Missing value'larni aniqlash va Median bilan to'ldirish
missing_count = df.isnull().sum()
med_maydon = df['maydon'].median()
df['maydon'] = df['maydon'].fillna(med_maydon)

# 3. Duplikatlarni o'chirish
df = df.drop_duplicates()

# 4. IQR yordamida outlier chegarasini hisoblash
Q1 = df['narx'].quantile(0.25)
Q3 = df['narx'].quantile(0.75)
IQR = Q3 - Q1
upper_bound = Q3 + 1.5 * IQR
clean_df = df[df['narx'] <= upper_bound]

clean_df
```
