```python
import numpy as np

w = np.array([2,-3])
b = -4
def neyron(x):
  z = np.dot(x,w) + b
  return int(z > 0 )    # int(True/False)ni Python 1 yoki 0 qilib qaytaradi

print(neyron([-5,1]))

#  -5 * 2 =-10.
#  1 * (-3) = -3.
#  -10+(-3) + (-4) = -17
#  -17 > 0 ? yoq => shu uchun False yani 0


print(neyron([5,1]))

```

```python
# AND || OR // NAND
X = np.array([[0,0],[0,1],[1,0],[1,1]])

def eshik(w,b):
  return (X @ np.array(w)+ b >0).astype(int)


print("And", eshik([2,2],-3))

# w = 2,2
# b = -3
# x =  [[0,0],[0,1],[1,0],[1,1]]

#  har bitta x ni w qatorga kopaytramiz

# [0,0] * [2,2] = [0,0] => 0+0 = 0
# [0,1] * [2,2] = [0,2] => 0+2 = 2
# [1,0] * [2,2] = [2,0] => 2+0 = 2
# [1,1] * [2,2] = [2,2] => 2+2 = 4
# w * x = [0,2,2,4]
# endi har birini B(bias)ga qoshib chiqamiz w*x +b
# [ 0,2,2,4] + -3 = [-3,-1,-1,1]

# >0)astype(int) endi buni shartga qoyib chiqadi
# 0>0 => False
# -1> 0 => False
# astype bolsa osha False 0 qiladi, Trueni 1 qilib natijani qaytaradi

```

```python

# OOP
class PerceptronNP:
  def __init__(self,lr=1.0,epochs=50):
    self.lr , self.epochs = lr, epochs
  def predict(self,X):
    return (X @ self.w + self.b > 0).astype(int)
  def fit(self,X,y):
    self.w , self.b  = np.zeros(X.shape[1]), 0.0 # w va b qiymatlarini beramiz
    for ep in range(1,self.epochs + 1): # nechda kiritgan bolsa shuncha qadam boladi,range ozi ohirgi soni olmaydi,shunga +1 qilinsa toliq qadamlarni oladi,masalani range ozi 49 olsa +1 bn 50 ham oladi
      xato = 0
      for xi,yi in zip(X,y):
        upd = self.lr* (yi - self.predict(xi))
        self.w, self.b = self.w + upd * xi, self.b + upd
        xato += int(upd != 0) #xato 0 dan kotta bolsa xato degan ozgaruvchiga + 1 qoshib ketadi
      if xato == 0:
        return ep

    return self.epochs



X = np.array([[0,0],[0,1],[1,0],[1,1]])

p =PerceptronNP()
print(p.fit(X, [0,0,0,1]))
print(p.w,p.b )
```
