```python
import matplotlib.pyplot as plt
import torch
import torch.nn.functional as F

x = torch.randn(1, 1, 28, 28)
# 1. Filtrlar
vertikal = torch.tensor([[-1.0, 0.0, 1.0], [-2.0, 0.0, 2.0], [-1.0, 0.0, 1.0]]).reshape(
    1, 1, 3, 3
)
gorizontal = vertikal.transpose(2, 3)
xiralash = torch.ones(1, 1, 3, 3) / 9
otkirlash = torch.tensor(
    [[0.0, -1.0, 0.0], [-1.0, 5.0, -1.0], [0.0, -1.0, 0.0]]
).reshape(1, 1, 3, 3)

filtrlar = {
    "vertikal": vertikal,
    "gorizontal": gorizontal,
    "xiralash": xiralash,
    "otkirlash": otkirlash,
}

# 2. Vizualizatsiya va TODO'larni to'ldirish
# (Faraz qilaylik, `x` rasmi allaqachon yuklangan [1, 1, H, W] tensor)
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
axes[0].imshow(x[0, 0], cmap="gray")
axes[0].set_title("asl")

for ax, (nom, f) in zip(axes[1:], filtrlar.items()):
    # TODO 1:
    natija = F.conv2d(x, f)
    # TODO 2:
    ax.imshow(natija[0, 0], cmap="gray")
    ax.set_title(nom)

for ax in axes:
    ax.axis("off")
plt.show()


```

```python
filtr_ident = torch.zeros(1, 1, 3, 3)
filtr_ident[0, 0, 1, 1] = 1.0

# Taxmin: Padding=1 bo'lgani uchun o'rtasidagi 1 piksel rasmni o'z holicha saqlaydi,
# natija asl `x` rasm bilan bir xil bo'ladi (Identity filter).
natija_ident = F.conv2d(x, filtr_ident, padding=1)

print("Bir xillikni tekshirish:", torch.allclose(natija_ident, x))
```
