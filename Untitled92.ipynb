```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider
import warnings
warnings.filterwarnings('ignore')

# 1. Ma'lumotlar bazasini yaratamiz (Ikki sinf: Moviy=0, Qizil=1)
X = np.array([[2, 2], [2, 3], [3, 2], [2.5, 1.5], [8, 7], [7, 8], [8, 8], [9, 7.5], [6.5, 6.5]])
y = np.array([0, 0, 0, 0, 1, 1, 1, 1, 1])
new_point = np.array([5.5, 5.0]) # Noma'lum yangi nuqta
k = 3 # K ta qo'shni

# Masofalarni oldindan hisoblab olamiz
euclid_dist = np.sqrt(np.sum((X - new_point)**2, axis=1))
manhat_dist = np.sum(np.abs(X - new_point), axis=1)

euclid_nearest = np.argsort(euclid_dist)[:k]
manhat_nearest = np.argsort(manhat_dist)[:k]

# Interaktiv chizish funksiyasi
def plot_knn_steps(Qadam):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

    # Asosiy dizaynni sozlash
    def setup_ax(ax, title):
        ax.set_title(title, fontsize=14, fontweight='bold', pad=15)
        ax.set_xlim(0, 10)
        ax.set_ylim(0, 10)
        ax.grid(True, linestyle='--', alpha=0.5)
        # Eski nuqtalarni chizish
        ax.scatter(X[y==0, 0], X[y==0, 1], c='dodgerblue', s=120, label="0-sinf (Moviy)", edgecolors='k')
        ax.scatter(X[y==1, 0], X[y==1, 1], c='crimson', s=120, label="1-sinf (Qizil)", edgecolors='k')
        ax.set_aspect('equal')

    setup_ax(ax1, "Euclidean (To'g'ri chiziq / Qush parvozi)")
    setup_ax(ax2, "Manhattan (90° burilish / Taksi yurishi)")

    # QADAM 1: Noma'lum nuqta paydo bo'ldi
    if Qadam >= 1:
        point_color = 'gold' # Boshida rangi noma'lum

        # QADAM 4: Ovoz berish jarayoni va Natija
        if Qadam == 4:
            euclid_votes = sum(y[euclid_nearest])
            point_color = 'crimson' if euclid_votes > k/2 else 'dodgerblue'

        ax1.scatter(new_point[0], new_point[1], c=point_color, marker='*', s=450, edgecolors='k', label="Yangi Nuqta")
        ax2.scatter(new_point[0], new_point[1], c=point_color, marker='*', s=450, edgecolors='k')

    # QADAM 2: Barcha nuqtalargacha masofa chizish
    if Qadam >= 2:
        for i in range(len(X)):
            # Evklid: To'g'ri diagonal chiziq
            ax1.plot([new_point[0], X[i, 0]], [new_point[1], X[i, 1]], 'gray', linestyle=':', alpha=0.4)
            # Manxetten: L-shaklidagi burchakli chiziqlar
            ax2.plot([new_point[0], X[i, 0]], [new_point[1], new_point[1]], 'gray', linestyle=':', alpha=0.4)
            ax2.plot([X[i, 0], X[i, 0]], [new_point[1], X[i, 1]], 'gray', linestyle=':', alpha=0.4)

    # QADAM 3: Eng yaqin K=3 ta qo'shnini qalin qilib ajratish
    if Qadam >= 3:
        for idx in euclid_nearest:
            ax1.plot([new_point[0], X[idx, 0]], [new_point[1], X[idx, 1]], 'limegreen', linewidth=2.5)
            ax1.scatter(X[idx, 0], X[idx, 1], facecolors='none', edgecolors='limegreen', s=300, linewidth=3)

        for idx in manhat_nearest:
            ax2.plot([new_point[0], X[idx, 0]], [new_point[1], new_point[1]], 'limegreen', linewidth=2.5)
            ax2.plot([X[idx, 0], X[idx, 0]], [new_point[1], X[idx, 1]], 'limegreen', linewidth=2.5)
            ax2.scatter(X[idx, 0], X[idx, 1], facecolors='none', edgecolors='limegreen', s=300, linewidth=3)

    if Qadam == 1: fig.suptitle("1-Qadam: Yangi noma'lum nuqta xaritaga tushdi", fontsize=16, y=1.02)
    elif Qadam == 2: fig.suptitle("2-Qadam: Barcha nuqtalargacha bo'lgan masofalar o'lchanadi", fontsize=16, y=1.02)
    elif Qadam == 3: fig.suptitle(f"3-Qadam: Eng qisqa masofali K={k} ta qo'shni topiladi", fontsize=16, y=1.02)
    elif Qadam == 4: fig.suptitle("4-Qadam: K ta qo'shni o'rtasida ovoz berilib, yakuniy rang aniqlanadi", fontsize=16, y=1.02, color='red')

    ax1.legend(loc='lower right')
    plt.show()

# Sliderni ishga tushirish (Slayderni surib bosqichlarni ko'ring)
interact(plot_knn_steps, Qadam=IntSlider(min=1, max=4, step=1, value=1, description='Qadam:'));
```

```python
import numpy as np

A = np.array([2, 3])
B = np.array([5, 7])

# Evklid masofasini hisoblash
euclidean_dist = np.sqrt(np.sum((A - B) ** 2))
# yoki: np.linalg.norm(A - B)

# Manxetten masofasini hisoblash
manhattan_dist = np.sum(np.abs(A - B))

print(f"Evklid masofasi: {euclidean_dist}")   # sqrt((3)^2 + (4)^2) = 5.0
print(f"Manxetten masofasi: {manhattan_dist}") # |3| + |4| = 7
```

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

# 1. Ma'lumotlarni yuklaymiz
X, y = load_iris(return_X_y=True)

# 2. Pipeline tuzamiz (StandardScaler + KNN)
# Eslatma: Pipeline ichidagi KNN modeliga 'knn' deb nom beramiz
pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', KNeighborsClassifier())
])

# 3. Grid (Parametrlar to'plami) tayyorlaymiz
# 'knn__' qo'shimchasi pipeline ichidagi 'knn' modelining parametriga murojaat qilish uchun kerak
grid = {
    'knn__n_neighbors': [1, 3, 5, 7, 9, 11, 15, 21],
    'knn__weights': ['uniform', 'distance'],
    'knn__metric': ['euclidean', 'manhattan']
}

# 4. Cross-validation obyektini belgilaymiz
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 5. GridSearchCV obyektini yaratamiz
gs = GridSearchCV(
    estimator=pipe,
    param_grid=grid,
    cv=cv,
    scoring='accuracy',
    n_jobs=-1
)

# 6. Qidiruvni ishga tushiramiz
gs.fit(X, y)

# 7. Natijalarni chiqaramiz
print("--- ENG YAXSHI NATIJALAR ---")
print("Eng yaxshi kombinatsiya (best_params_):", gs.best_params_)
print(f"Eng yaxshi Cross-Validation balli (best_score_): {gs.best_score_:.4f}")

# Istasangiz eng yaxshi modelni to'g'ridan-to'g'ri ishlatishingiz mumkin:
best_model = gs.best_estimator_
```
