```python
import numpy as np
```

```python
X = np.array([[0.5,0.8]])
```

```python
W1 = np.array([
    [0.2, 0.5,-0.1],
    [0.4,-0.3,-0.6]
])

b1 = np.array([[0.1,0.1, 0.1]])

b1.shape
```

```python
W2 = np.array([
    [0.7],
    [0.2],
    [-0.4]
])

b2 = np.array([[0.2]])

W2.shape
```

```python
def relu(z):
  return np.maximum(0,z)

def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
# Forward pass jarayoni

Z1 = np.dot(X,W1)+b1
print(f"1-qatlam chiziqli yig'indisi (Z1):\n{Z1}\n")
A1 = relu(Z1)
print(f"1-qatlam chiqishi (A1 - ReLU dan keyin):\n{A1}\n")
Z2 = np.dot(A1,W2)+b2
print(f"2-qatlam chiziqli yig'indisi (Z2):\n{Z2}\n")
Y_pred = sigmoid(Z2)
print(f"Bashorat natijasi : {Y_pred}")
```

```python
X = 2
w = 1.5
b = -1

def sigmoid(z):
  return 1/(1+np.exp(-z))

def tanh(z):
   return (np.exp(z) - np.exp(-z))/(np.exp(z) + np.exp(-z))

Z = X * w + b

k_sigmoid = sigmoid(Z)
k_tanh = tanh(Z)

# hosilasini hisoblaymiz

def der_sigmoid(k):
    return k*(1-k)
def der_tanh(k):
    return 1 - k**2
sigmoid_hosila = der_sigmoid(k_sigmoid)
tanh_hosila = der_tanh(k_tanh)
print(f"Sigmoidni hosilasi : {sigmoid_hosila}")
print(f"Tanhning hosilasi: {tanh_hosila}")
```
