```python
w = np.array([2,-3])
b = -4

def neyron(x):
  z = np.dot(x,w)+b
  return int(z>0)

print(neyron([5,1]))
print(neyron([3,2]))
```

```python
import numpy as np


class PerceptronNP:

    def __init__(self, lr=0.1, epochs=50):
        # Har bir parametrni alohida qatorda belgilash xavfsizroq
        self.lr = lr
        self.epochs = epochs
        self.w = None
        self.b = None

    def predict(self, X):
        return (X @ self.w + self.b > 0).astype(int)

    def fit(self, X, y):
        self.w = np.zeros(X.shape[1])
        self.b = 0.0

        for ep in range(1, self.epochs + 1):
            xato = 0
            for xi, yi in zip(X, y):
                upd = self.lr * (yi - self.predict(xi))
                self.w = self.w + upd * xi
                self.b = self.b + upd
                if upd != 0:
                    xato += 1

            # `if xato == 0` sharti for xi, yi siklidan TASHQARIDA (bitta surilgan) bo'lishi kerak!
            if xato == 0:
                return ep

        return self.epochs


# Sinab ko'rish
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

p = PerceptronNP()
print("To'liq o'rganilgan epoch:", p.fit(X, [0, 0, 0, 1]))
print("Vaznlar (w) va Bias (b):", p.w, p.b)
```
