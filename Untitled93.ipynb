Bu Telco Customer Churn (Telekommunikatsiya mijozlarining ketib qolishi) datasetining ustunlari. Har bir ustunning nimani anglatishi bilan tanishib chiqing:


1. customerID — Mijozning unikal ID raqami/kodi. ML model uchun hech qanday mantiqiy qiymatga ega emas, Feature Engineering bosqichida tushirib qoldiriladi (drop).

2. gender — Mijozning jinsi (Male/Female).

3. eniorCitizen — Mijozning keksalar (pensionerlar) yoshidami-yo'qligi. Qiymati: 1 (ha) yoki 0 (yo'q).

4. Partner — Mijozning turmush o'rtog'i/sherigi bormi-yo'qligi (Yes/No).
5. Dependents — Mijozning qaramog'ida odamlar (farzand, ota-ona va h.k.) bormi-yo'qligi (Yes/No).
6. tenure — Mijoz bu kompaniya xizmatlaridan qancha oydan beri foydalanib kelayotgani (oylar soni).
7. PhoneService — Mijozda telefon xizmati ulanganmi-yo'qligi (Yes/No).
8. MultipleLines — Mijozda bir nechta telefon liniyasi bormi (Yes, No, No phone service).
9. InternetService — Mijoz foydalanadigan internet turi (DSL, Fiber optic, No).
10. OnlineSecurity — Onlayn xavfsizlik (antivirus/firewall) xizmati ulanganmi (Yes, No, No internet service).
11. OnlineBackup — Ma'lumotlarni bulutli saqlash (backup) xizmati ulanganmi 12. (Yes, No, No internet service).
13. DeviceProtection — Qurilmalarni sug'urtalash/himoyalash xizmati ulanganmi (Yes, No, No internet service).
14. TechSupport — Texnik qo'llab-quvvatlash xizmati ulanganmi (Yes, No, No internet service).
15. StreamingTV — Onlayn TV/televideniye xizmati ulanganmi (Yes, No, No internet service).
16. StreamingMovies — Onlayn kinolar tomosha qilish xizmati ulanganmi (Yes, No, No internet service).
17. Contract — Shartnoma turi (Month-to-month — oylik, One year — 1 yillik, Two year — 2 yillik).
18. PaperlessBilling — Qog'ozsiz (elektron) hisob-faktura xizmati ulanganmi (Yes/No).
19. PaymentMethod — To'lov usuli (Electronic check, Mailed check, Bank transfer, Credit card).
20. MonthlyCharges — Mijoz har oyda to'laydigan summasi (sonli ko'rinishda).
21. TotalCharges — Mijoz butun foydalanish davri mobaynida to'lagan umumiy summasi.


Churn — Target (Maqsadli ustun). Mijoz kompaniyadan ketib qoldimi yoki yo'qmi (Yes — ketgan, No — qolgan).

```python
mask = df["TotalCharges"].str.strip() == ''
df.loc[mask,'TotalCharges'] = np.nan
df = df.dropna(subset=['TotalCharges'])
print(df.shape)
# 7043 dan 7032ta qatorga tushdi
```

```python
df["MonthlyCharges"].groupby(df["Churn"]).mean()
```

```python
df.groupby(["MonthlyCharges"])['Churn'].value_counts().sort_values(ascending=False)
```

```python
from sklearn.preprocessing import LabelEncoder
binary_cols = ['Churn', 'gender', 'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling']

le = LabelEncoder()
for l in binary_cols:
    df[l] = le.fit_transform(df[l])
df
```

```python
col = ["MultipleLines", "OnlineSecurity", "OnlineBackup", "DeviceProtection", "TechSupport", "StreamingTV", "StreamingMovies"]

for c in col:
    df[c] = df[c].replace({"No": 0, "Yes": 1, "No phone service": 0, "No internet service": 0})

df
```

```python
df["Contract"].value_counts()
```

```python
multi_cols = ["Contract","InternetService","PaymentMethod"]
df = pd.get_dummies(df, columns=multi_cols)
df
```
