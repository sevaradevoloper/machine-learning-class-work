```python
# formulalarni funksiya qilamiz
def linear_params(n_in,n_out):
  return n_in * n_out + n_out
def convolution_params(k ,c_in , c_out):
  return k * k * c_in * c_out + c_out
# Formula 1 (n_in × n_out + n_out).
# Formula 2 (K × K × C_in × C_out + C_out).

# n_in: kirishlar soni
# n_out: neyronlar soni
# P: jami parametr


# K: filtr tomoni (K = 3 bo'lsa filtr 3×3)
# C_in: kirishdagi kanallar soni (kulrang 1, rangli 3)
# C_out: filtrlar soni
```

```python
# tekshiramiz
print(linear_params(64,32), convolution_params(3,1,8))
```

```python
for h, w, c, n_out, filters in [(8, 8, 1, 32, 8),
                                (28, 28, 1, 128, 32),
                                (224, 224, 3, 1000, 64)]:
    lin = linear_params(h * w * c, n_out)
    cnv = convolution_params(3, c, filters)
    print(h, w, c, lin, cnv, round(lin / cnv))
```

```python
import numpy as np
from sklearn.datasets import load_digits


class DigitsData:
    def __init__(self):
        digits = load_digits()
        self.X = digits.images
        self.y = digits.target

    def info(self):
        n, h, w = self.X.shape
        return {
            "rasmlar": n,
            "balandlik": h,
            "kenglik": w,
            "kanal": 1,
            "sinflar": len(np.unique(self.y)),
            "min": float(self.X.min()),
            "max": float(self.X.max()),
        }

    def class_counts(self):
        return np.bincount(self.y)

    def get_image(self, i):
        return self.X[i].astype(int), int(self.y[i])
```

```python
class ParamCounter:
    @staticmethod
    def linear(n_in, n_out):
        return n_in * n_out + n_out

    @staticmethod
    def conv(k, c_in, c_out):
        return k * k * c_in * c_out + c_out

    def compare(self, h, w, c, neurons, filters, k=3):
        lin = self.linear(h * w * c, neurons)
        cnv = self.conv(k, c, filters)
        return {"linear": lin, "conv": cnv, "farq": round(lin / cnv)}
```

```python
class Experiment:
    def __init__(self, data, counter):
        self.data = data
        self.counter = counter

    def run(self, neurons=32, filters=8):
        info = self.data.info()
        return self.counter.compare(
            info["balandlik"], info["kenglik"], info["kanal"],
            neurons, filters)
```

```python
data = DigitsData()
print(data.info())
print(data.class_counts())
```

```python
image, label = data.get_image(0)
print(image)
print(label)

counter = ParamCounter()
print(counter.compare(28, 28, 1, 128, 32))

exp = Experiment(data, counter)
print(exp.run())
```
