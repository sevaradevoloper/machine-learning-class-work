```python
import numpy as np
from sklearn.datasets import load_digits
digits = load_digits()
```

```python
X , y = digits.images, digits.target
```

```python
print(X.shape, X.min(), X.max(), y[0])
# (1797, 8, 8) 0.0 16.0 0 => 1797 ta rasm, 8 qator, 8 ustun
# X.min() → 0.0: eng kichik piksel qiymati (oq).
# X.max() → 16.0: eng katta piksel qiymati (qora).
# y[0] → 0: birinchi rasmning to'g'ri javobi.
```

```python
print(X[0].astype(int))
# X[0]: birinchi rasm (8×8 jadval).
# .astype(int): sonlarni butun songa aylantiradi. Aks holda 5. 13. kabi nuqtali ko'rinadi (sonlar aslida float). Bu faqat chiroyli ko'rinish uchun, qiymat o'zgarmaydi.
```

```python
print(np.bincount(y))
# np.bincount(y) har bir raqam (0, 1, 2, ... 9) necha marta uchraganini sanaydi. Masalan "0" raqamidan 178 ta rasm bor, "1" dan 182 ta. Sinflar taxminan teng, bu o'qitish uchun yaxshi.
```
