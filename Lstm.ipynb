```python
import re

def tokenize(text):
    # Bu yerda text.lower() qavsi yopilganiga e'tibor bering
    return re.findall(r'[a-zA-Z0-9]+', str(text).lower())

# Butun df ni emas, balki uning ichidagi 'text' ustunini tokenizatsiya qilamiz:
df['tokenized_text'] = df['text'].apply(tokenize)

# Natijani tekshirish uchun birinchi 5 ta qatorni chiqaramiz:
print(df['tokenized_text'].head())
```

```python
from collections import Counter
from sklearn.model_selection import train_test_split

tr, va = train_test_split(df, test_size=0.2,
                         stratify=df.label, random_state=0)

counts = Counter(w for t in tr.text for w in tokenize(t))
vocab = {"<pad>": 0, "<unk>": 1}


for w, c in counts.items():
    if c >= 2:
        vocab[w] = len(vocab)
print(len(tr), len(va), len(vocab))
print(counts.most_common(4))
```

```python
def encode(text, maxlen=30):
    ids = [vocab.get(w, 1) for w in tokenize(text)][:maxlen]
    return [0] * (maxlen - len(ids)) + ids

print(encode("Ok lar... Joking wif u oni...", 10))
```
